# 🎬 Netflix Data Analysis Project

## 📌 Project Objective
The objective of this project is to perform Exploratory Data Analysis (EDA) on the Netflix dataset to understand content distribution, trends, genres, ratings, and duration using Python, NumPy, Pandas, Matplotlib, and Seaborn.


## 1️⃣ Import Libraries

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns

sns.set_style("whitegrid")
```

## 2️⃣ Load Dataset

```python
df = pd.read_csv("netflix_titles.csv")
df.head()
```

## 3️⃣ Data Understanding
- Checking shape of dataset
- Viewing columns
- Checking data types
- Identifying missing values


```python
df.shape 
df.columns
df.info() 
df.isnull()
```

## 4️⃣ Data Cleaning
Handling missing values and preparing dataset for analysis.

```python
df['director'].fillna("Unkown")
df['cast'].fillna("Not available")
df['country'].fillna("Unknown")
df.dropna(subset=['rating'])
```

## 5️⃣ Feature Engineering
Extracting useful information from existing columns.

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), format='%B %d, %Y', errors='coerce')
```

```python
df.head()
```

```python
df['release_year'] = df['date_added'].dt.year
```

```python
df.head()
```

```python
df['type'].value_counts()
```

## 6️⃣ Exploratory Data Analysis (EDA)


### 6.1 Movies vs TV Shows


```python
sns.countplot(x='type',data=df)
plt.title("Movies vs TV Shows") 
plt.show()
```

### 6.2 Percentage of Movies


```python
total = len(df)
movies_count = len(df[df['type'] == 'Movie'])

movie_percentage = (movies_count / total)*100 
print("Percentage of movies:",movie_percentage)
```

### 6.3 Top 10 Countries


```python
top_countries = df['country'].value_counts().head(10)

top_countries.plot(kind='bar')
plt.title("TOP 10 Countris Producing Content")
plt.xticks(rotation=45)
plt.show()
```

### 6.4 Content Growth Over Years


```python
df['year_added'].value_counts().sort_index().plot(kind='line')
plt.title("Content Added Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Shows ")
plt.show()
```

### 6.5 Top 10 Genres


```python
genres = df['listed_in'].str.split(',').explode() 
top_genres = genres.value_counts().head(10)
top_genres.plot(kind='bar')
plt.title("Top 10 Genres")
plt.show()
```

### 6.6 Rating Distribution


```python
sns.countplot(y='rating',data=df,order=df['rating'].value_counts().index) 
plt.title("Rating Distribution")
plt.show()
```

## 7️⃣ Movie Duration Analysis


```python
df['duration_num'] = df['duration'].str.extract('(\d+)') 
df['duration_num'] = pd.to_numeric(df['duration_num']) 

movies = df[df['type'] == 'Movie'] 

average_duration = np.mean(movies['duration_num'])
print("Average Movie Duration:", average_duration)
```

```python
movies.sort_values(by='duration_num',ascending=False).head(5)
```

```python
df['month_added'].value_counts().sort_index().plot(kind='bar')
plt.title("Monthly Content Added")
plt.xlabel("Month")
plt.ylabel("Count")
plt.show()
```
